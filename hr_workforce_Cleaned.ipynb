## -Path Validation

```python
from pathlib import Path

data_path = Path("Final HR")

try:
    if not data_path.exists():
        raise FileNotFoundError("Data folder does not exist.")

    if not data_path.is_dir():
        raise NotADirectoryError("The path is not a directory.")

    csv_files = list(data_path.glob("*.csv"))

    if not csv_files:
        raise FileNotFoundError("No CSV files found.")

    print("Folder:", data_path.resolve())
    print("Number of CSV files:", len(csv_files))

except (FileNotFoundError, NotADirectoryError) as e:
    print("Path Error:", e)
```

## -Data Loading

```python
tables = {}
failed_files = {}

for file in csv_files:
    try:
        df = pd.read_csv(file)
        tables[file.stem.strip()] = df
        print(f"Loaded successfully: {file.name}")

    except (pd.errors.ParserError, UnicodeDecodeError, OSError, ValueError) as e:
        failed_files[file.name] = str(e)

        print(f"Failed to load: {file.name}")
        print("Reason:", e)

print("\nLoaded tables:", len(tables))
print("Failed files:", len(failed_files))
```

# 02-Data Cleaning & Preprocessing

## -Create a backup of original data

```python
original_tables = {
    name: df.copy(deep=True)
    for name, df in tables.items()
}

clean_tables = {
    name: df.copy(deep=True)
    for name, df in tables.items()
}

print("Original data backed up successfully!")
print("Number of tables:", len(clean_tables))
```

## -Data Inspection

```python
for name, df in clean_tables.items():
    print(f"\n{'='*50}")
    print(f"Table: {name}")
    print("Shape:", df.shape)
    print("\nData Types:")
    print(df.dtypes)
    print("\nColumns:")
    print(df.columns.tolist())
    df.info()
```

## -Fix IssuesText
    ### Remove extra spaces
    ### Convert empty strings to NaN

```python
for name, df in clean_tables.items():

    text_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in text_columns:

        # Remove extra spaces
        df[col] = df[col].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

        # Convert empty strings to NaN
        df[col] = df[col].replace(
            r"^\s*$", np.nan, regex=True
        )

print("Text issues fixed successfully!")
```

## -Missing Values Analysis

```python
for name, df in clean_tables.items():
    print(f"\nMissing Values in {name}:")

    missing = pd.DataFrame({
        "Missing Count": df.isnull().sum(),
        "Missing Percentage": (
            df.isnull().mean() * 100
        ).round(2)
    })

    display(missing[missing["Missing Count"] > 0])
```

```python
df = clean_tables["Fact_PerformanceReview"]

print("Missing IDs Before:", df["PerformanceID"].isna().sum())

df = df.dropna(subset=["PerformanceID"])

clean_tables["Fact_PerformanceReview"] = df

print("Missing IDs After:", df["PerformanceID"].isna().sum())
print("Rows After Cleaning:", df.shape[0])
```

## -Remove Duplicates

```python
for name, df in clean_tables.items():
    print(f"\nTable: {name}")

    print("Duplicate Rows:", df.duplicated().sum())

    df.drop_duplicates(inplace=True)

    print("Shape After Removing Duplicates:", df.shape)
```

## -Convert Dates

```python
df = clean_tables["Dim_Date (1)"]

missing_before = df["Date"].isna().sum()

df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)

missing_after = df["Date"].isna().sum()

print("Invalid Dates:", missing_after - missing_before)
print("Date Conversion Completed!")
```

```python
employee = clean_tables["Dim_Employee (1)"]

relationships = {
    "EducationLevelID": "Dim_EducationLevel",
    "DepartmentID": "Dim_Department (1)",
    "JobRoleID": "Dim_JobRole (1)",
    "StateID": "Dim_State (1)"
}

for fk, dim_table in relationships.items():

    valid_ids = clean_tables[dim_table][
        fk
    ] if fk in clean_tables[dim_table].columns else None
```

```python
fact = clean_tables["Fact_PerformanceReview"]

relationships = {
    "EmployeeID": ("Dim_Employee (1)", "EmployeeID"),
    "ReviewDateKey": ("Dim_Date (1)", "DateKey")
}

for fk, (table, pk) in relationships.items():

    valid_ids = clean_tables[table][pk]

    missing = fact[fk].isna().sum()

    invalid = (
        fact[fk].notna() &
        ~fact[fk].isin(valid_ids)
    ).sum()

    print(f"\n{fk}")
    print("Missing References:", missing)
    print("Invalid References:", invalid)
```

```python
primary_keys = {
    "Fact_PerformanceReview": "PerformanceID",
    "Dim_Employee (1)": "EmployeeID",
    "Dim_EducationLevel": "EducationLevelID",
    "Dim_State (1)": "StateID",
    "Dim_Date (1)": "DateKey",
    "Dim_Department (1)": "DepartmentID",
    "Dim_JobRole (1)": "JobRoleID",
    "Dim_SatisfactionLevel": "SatisfactionID",
    "Dim_RatingLevel": "RatingID"
}

for table, pk in primary_keys.items():

    df = clean_tables[table]

    print(f"\n{table}")

    print("Missing IDs:", df[pk].isna().sum())
    print("Duplicate IDs:", df[pk].duplicated().sum())
```

## -Logical Checks

```python
## Employee Table
df = clean_tables["Dim_Employee (1)"]

print("Invalid Age:")
display(df[(df["Age"] <= 0) | (df["Age"] > 100)])

print("Negative Salary:")
display(df[df["Salary"] < 0])

print("Invalid Years:")
display(df[df["YearsInMostRecentRole"] > df["YearsAtCompany"]])
```

```python
## Performance Table
df = clean_tables["Fact_PerformanceReview"]

rating_columns = [
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
    "SelfRating",
    "ManagerRating"
]

for col in rating_columns:

    invalid = df[
        ~df[col].between(1, 5) & df[col].notna()
    ]

    print(f"{col} - Invalid Values: {len(invalid)}")
```

```python
## Training Validation
df = clean_tables["Fact_PerformanceReview"]

invalid_training = df[
    df["TrainingOpportunitiesTaken"] >
    df["TrainingOpportunitiesWithinYear"]
]

print("Invalid Training Records:")
display(invalid_training)
```
